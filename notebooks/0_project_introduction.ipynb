### Problem Statement:

- Insurance companies suffer huge financial losses due to fraudulent vehicle
insurance claims.
- Traditional fraud detection methods often:
    - are manual
    - consume time
    - increase operational cost
    - fail to detect hidden fraud patterns
- This can be due to:
    - increase claim volume
    - complex fraud behavior
    - overlapping fraud indicators
- So, there is a need for an intelligent automated fraud detection system.

### Project Objectives:

##### Primary Objectives:
- Detect fraudulent insurance claims using Machine Learning
- Analyze hidden fraud patterns from insurance data
- Build an intelligent fraud prediction system
- Reduce financial losses caused by false claims

##### Technical Objectives:
- Perform EDA and fraud analysis
- Handle imbalance dataset
- Apply feature engineering
- Train multiple ML models
- Optimize model performance using hyperparameter tuning
- Apply threshold tuning to adjust fraud detection sensitivity
- Evaluate model using fraud-oriented metrics
- Build an end-to-end web app to display results
